```python
# CELL 2: Simplified Path Setup
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
DATA_DIR = PROJECT_ROOT / "data"

# Only create paths we actually need
DATA_RAW = DATA_DIR / "raw" / "OriginalImages"
DATA_LANDMARKS = DATA_DIR / "landmarks"
DATA_ALIGNED = DATA_DIR / "aligned"

# Create only essential directories
DATA_LANDMARKS.mkdir(parents=True, exist_ok=True)
DATA_ALIGNED.mkdir(parents=True, exist_ok=True)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DATA_RAW:", DATA_RAW)
print("DATA_LANDMARKS:", DATA_LANDMARKS)
print("DATA_ALIGNED:", DATA_ALIGNED)
print("\n✓ Path setup complete")
```

```python
# CELL 3: StyleGAN2 Setup (Run once)
import subprocess
from pathlib import Path

# Define paths
STYLEGAN2_DIR = PROJECT_ROOT / "stylegan2-ada-pytorch"
MODELS_DIR = PROJECT_ROOT / "models" / "pretrained"
MODELS_DIR.mkdir(parents=True, exist_ok=True)

print("="*60)
print("STYLEGAN2 SETUP")
print("="*60)

# Clone StyleGAN2 if not exists
if not STYLEGAN2_DIR.exists():
    print("\n📥 Cloning StyleGAN2-ADA-PyTorch...")
    subprocess.run([
        "git", "clone",
        "https://github.com/NVlabs/stylegan2-ada-pytorch.git",
        str(STYLEGAN2_DIR)
    ])
    print("✅ Cloned successfully!")
else:
    print("✅ StyleGAN2 already cloned")

# Add to Python path
import sys
if str(STYLEGAN2_DIR) not in sys.path:
    sys.path.insert(0, str(STYLEGAN2_DIR))
    print(f"✅ Added to sys.path: {STYLEGAN2_DIR}")

# Download pre-trained FFHQ model if not exists
FFHQ_MODEL = MODELS_DIR / "ffhq.pkl"
if not FFHQ_MODEL.exists():
    print("\n📥 Downloading pre-trained FFHQ model (takes a few minutes)...")
    import urllib.request
    url = "https://nvlabs-fi-cdn.nvidia.com/stylegan2-ada-pytorch/pretrained/ffhq.pkl"
    urllib.request.urlretrieve(url, str(FFHQ_MODEL))
    print(f"✅ Downloaded to: {FFHQ_MODEL}")
else:
    print(f"✅ FFHQ model exists: {FFHQ_MODEL}")

print("\n" + "="*60)
print("✅ StyleGAN2 setup complete!")
print("="*60)
```

```python
# CELL 4: Prepare Caricature Dataset for StyleGAN2 Training
import shutil
from pathlib import Path
from PIL import Image

# Target resolution for StyleGAN2
TARGET_RESOLUTION = 512  # Can be 256, 512, or 1024

STYLEGAN_DATA_DIR = PROJECT_ROOT / "data" / "stylegan_training"
STYLEGAN_DATA_DIR.mkdir(parents=True, exist_ok=True)

print("="*60)
print("PREPARING DATASET FOR STYLEGAN2")
print("="*60)

# Collect all caricature images (files starting with 'c') from all actor folders
print(f"\n📂 Scanning for caricatures in: {DATA_RAW}")
print(f"   Looking for files starting with 'c' in subdirectories...")

caricature_files = []
actor_folders = [d for d in DATA_RAW.iterdir() if d.is_dir()]

for actor_folder in actor_folders:
    # Find all files starting with 'c' (caricatures)
    carics = list(actor_folder.glob("c*.*"))
    caricature_files.extend(carics)
    if carics:
        print(f"   Found {len(carics)} caricature(s) in: {actor_folder.name}")

print(f"\n📊 Total caricatures found: {len(caricature_files)}")

if len(caricature_files) == 0:
    print("❌ No caricature images found!")
    print("   Make sure files starting with 'c' exist in actor folders")
else:
    # Resize and copy to training folder
    training_imgs_dir = STYLEGAN_DATA_DIR / "images"
    training_imgs_dir.mkdir(exist_ok=True)
    
    print(f"\n🔄 Resizing all images to {TARGET_RESOLUTION}x{TARGET_RESOLUTION}...")
    
    successful = 0
    for i, img_path in enumerate(caricature_files):
        try:
            # Load, resize, and save
            img = Image.open(img_path).convert('RGB')
            img_resized = img.resize((TARGET_RESOLUTION, TARGET_RESOLUTION), Image.LANCZOS)
            
            # Save with sequential naming
            dest = training_imgs_dir / f"img{i:05d}.png"
            img_resized.save(dest, 'PNG')
            successful += 1
        except Exception as e:
            print(f"   ⚠️  Failed to process {img_path.name}: {e}")
    
    print(f"✅ Successfully processed {successful}/{len(caricature_files)} images")
    print(f"💾 Saved to: {training_imgs_dir}")
    
    # Create dataset.zip for StyleGAN2
    print(f"\n📦 Creating dataset.zip for StyleGAN2...")
    import sys
    sys.path.insert(0, str(STYLEGAN2_DIR))
    
    # Run StyleGAN2's dataset_tool
    subprocess.run([
        sys.executable,
        str(STYLEGAN2_DIR / "dataset_tool.py"),
        "--source", str(training_imgs_dir),
        "--dest", str(STYLEGAN_DATA_DIR / "caricatures.zip"),
        "--resolution", f"{TARGET_RESOLUTION}x{TARGET_RESOLUTION}"
    ])
    
    DATASET_ZIP = STYLEGAN_DATA_DIR / "caricatures.zip"
    if DATASET_ZIP.exists():
        print(f"✅ Dataset created: {DATASET_ZIP}")
        print(f"   Size: {DATASET_ZIP.stat().st_size / 1024 / 1024:.2f} MB")
    else:
        print("❌ Dataset creation failed!")

print("\n" + "="*60)
print("✅ Dataset preparation complete!")
print("="*60)
```

```python
# CELL 5: Fine-tune StyleGAN2 (This will take time - GPU recommended)
import subprocess
from pathlib import Path

TRAINING_DIR = PROJECT_ROOT / "models" / "training_runs"
TRAINING_DIR.mkdir(parents=True, exist_ok=True)

print("="*60)
print("FINE-TUNING STYLEGAN2 ON CARICATURES")
print("="*60)

# Check if dataset exists
DATASET_ZIP = STYLEGAN_DATA_DIR / "caricatures.zip"
if not DATASET_ZIP.exists():
    print("❌ Dataset not found! Please run Cell 4 first.")
else:
    # Training configuration
    TRAINING_CONFIG = {
        'data': str(DATASET_ZIP),
        'resume': str(FFHQ_MODEL),  # Start from pre-trained FFHQ
        'outdir': str(TRAINING_DIR),
        'gpus': 1,  # Change based on your GPU count
        'kimg': 1000,  # Train for 1000k images (adjust based on dataset size)
        'snap': 10,  # Save checkpoint every 10 ticks
        'metrics': 'fid50k_full',  # Evaluate quality
        'mirror': 1,  # Use horizontal flipping
        'aug': 'ada',  # Use adaptive augmentation for small datasets
    }
    
    # Estimate training time
    num_caricatures = len(list((STYLEGAN_DATA_DIR / "images").glob("*.png")))
    print(f"\n📊 Training info:")
    print(f"   Dataset size: {num_caricatures} images")
    print(f"   Resolution: {TARGET_RESOLUTION}x{TARGET_RESOLUTION}")
    print(f"   Starting from: FFHQ pre-trained model")
    print(f"   Training iterations: {TRAINING_CONFIG['kimg']}k images")
    print(f"   GPU count: {TRAINING_CONFIG['gpus']}")
    print(f"   Augmentation: {TRAINING_CONFIG['aug']}")
    print(f"   Estimated time: ~2-4 hours on RTX 3090 (depends on GPU)")
    
    proceed = input("\n⚠️  Start training? (y/n): ")
    
    if proceed.lower() == 'y':
        print("\n🚀 Starting training...")
        
        # Build command
        cmd = [
            sys.executable,
            str(STYLEGAN2_DIR / "train.py"),
        ]
        for key, value in TRAINING_CONFIG.items():
            cmd.extend([f"--{key}", str(value)])
        
        # Run training
        subprocess.run(cmd, cwd=str(STYLEGAN2_DIR))
        
        print("\n✅ Training complete!")
        print(f"📁 Models saved in: {TRAINING_DIR}")
    else:
        print("⏸️  Training cancelled")

print("="*60)
```

```python
# CELL 6: StyleGAN2 Inference - Style Transfer
import torch
import pickle
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

print("="*60)
print("STYLEGAN2 INFERENCE - STYLE TRANSFER")
print("="*60)

# Find latest trained model
training_runs = sorted(TRAINING_DIR.glob("*-caricatures"))
if not training_runs:
    print("❌ No trained models found!")
    print("   Using pre-trained FFHQ for demo...")
    model_path = FFHQ_MODEL
else:
    latest_run = training_runs[-1]
    # Find latest snapshot
    snapshots = sorted(latest_run.glob("network-snapshot-*.pkl"))
    model_path = snapshots[-1] if snapshots else FFHQ_MODEL

print(f"✅ Using model: {model_path}")

# Load StyleGAN2 Generator
print("\n⏳ Loading StyleGAN2 generator...")
with open(model_path, 'rb') as f:
    G = pickle.load(f)['G_ema'].cuda() if torch.cuda.is_available() else pickle.load(f)['G_ema']

print("✅ Generator loaded!")
print(f"   Latent dim: {G.z_dim}")
print(f"   Output resolution: {G.img_resolution}x{G.img_resolution}")

# Function to project image to latent space
def project_image_to_latent(img_path, G, num_steps=1000):
    """
    Project an image to StyleGAN2's W+ latent space.
    This is the encoder part - maps your image to latent code.
    """
    from torchvision import transforms
    
    # Load and preprocess image
    img = Image.open(img_path).convert('RGB').resize((G.img_resolution, G.img_resolution))
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
    ])
    target = transform(img).unsqueeze(0)
    
    if torch.cuda.is_available():
        target = target.cuda()
    
    # Initialize latent code
    w_avg = G.mapping.w_avg
    w = w_avg.clone().unsqueeze(0).repeat(1, G.mapping.num_ws, 1)
    w.requires_grad = True
    
    # Optimizer
    optimizer = torch.optim.Adam([w], lr=0.01)
    
    # Projection loop
    print(f"   Projecting image to latent space ({num_steps} steps)...")
    for step in range(num_steps):
        # Generate image from latent
        synth_img = G.synthesis(w, noise_mode='const')
        
        # Loss: pixel difference
        loss = torch.nn.functional.mse_loss(synth_img, target)
        
        # Optimize
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        if step % 100 == 0:
            print(f"   Step {step}/{num_steps}, Loss: {loss.item():.4f}")
    
    print("   ✅ Projection complete!")
    return w.detach()

# Collect caricature images for testing
caricature_files = []
actor_folders = [d for d in DATA_RAW.iterdir() if d.is_dir()]
for actor_folder in actor_folders:
    carics = list(actor_folder.glob("c*.*"))
    caricature_files.extend(carics)

if caricature_files:
    demo_img_path = caricature_files[0]
    print(f"\n🎨 Testing style transfer on: {demo_img_path.name}")
    print(f"   From folder: {demo_img_path.parent.name}")
    
    # Project to latent space
    latent_code = project_image_to_latent(demo_img_path, G, num_steps=500)
    
    # Generate stylized caricature
    with torch.no_grad():
        stylized = G.synthesis(latent_code, noise_mode='const')
    
    # Convert to image
    stylized_img = (stylized.permute(0, 2, 3, 1) * 127.5 + 128).clamp(0, 255).to(torch.uint8)
    stylized_np = stylized_img[0].cpu().numpy()
    
    # Load original for comparison
    original_img = Image.open(demo_img_path).convert('RGB').resize((G.img_resolution, G.img_resolution))
    original_np = np.array(original_img)
    
    # Save result
    OUTPUT_DIR = PROJECT_ROOT / "data" / "stylegan_results"
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    output_stylized = OUTPUT_DIR / f"{demo_img_path.stem}_stylized.png"
    Image.fromarray(stylized_np).save(output_stylized)
    
    print(f"✅ Style transfer complete!")
    print(f"💾 Saved: {output_stylized}")
    
    # Visualize
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    axes[0].imshow(original_np)
    axes[0].set_title('Original\nCaricature', fontweight='bold')
    axes[0].axis('off')
    
    axes[1].imshow(stylized_np)
    axes[1].set_title('StyleGAN2\nStylized', fontweight='bold', color='#5cb85c')
    axes[1].axis('off')
    
    # Show difference
    axes[2].imshow(np.abs(original_np.astype(float) - stylized_np.astype(float)).astype(np.uint8))
    axes[2].set_title('Difference\n(Style Changes)', fontweight='bold', color='#d9534f')
    axes[2].axis('off')
    
    plt.suptitle('StyleGAN2 Style Transfer Test', fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()
else:
    print("⚠️  No caricature images found (files starting with 'c')")

print("="*60)
```

```python
# CELL 7: Batch Style Transfer on All Caricatures
from tqdm import tqdm

print("="*60)
print("BATCH STYLE TRANSFER")
print("="*60)

# Get all caricatures from all actor folders
caricature_files = []
actor_folders = [d for d in DATA_RAW.iterdir() if d.is_dir()]
for actor_folder in actor_folders:
    carics = list(actor_folder.glob("c*.*"))
    caricature_files.extend(carics)

print(f"Found {len(caricature_files)} caricatures to process")

if len(caricature_files) == 0:
    print("❌ No caricature images found!")
else:
    # Output directory for stylized results
    OUTPUT_STYLIZED = PROJECT_ROOT / "data" / "stylegan_stylized"
    OUTPUT_STYLIZED.mkdir(parents=True, exist_ok=True)
    
    print(f"\n🎨 Processing {len(caricature_files)} images...")
    print(f"💾 Output directory: {OUTPUT_STYLIZED}")
    
    for img_path in tqdm(caricature_files, desc="Stylizing"):
        try:
            # Project and generate
            latent = project_image_to_latent(img_path, G, num_steps=300)  # Fewer steps for batch
            
            with torch.no_grad():
                stylized = G.synthesis(latent, noise_mode='const')
            
            # Save
            stylized_img = (stylized.permute(0, 2, 3, 1) * 127.5 + 128).clamp(0, 255).to(torch.uint8)
            stylized_np = stylized_img[0].cpu().numpy()
            
            # Include actor name in output filename
            actor_name = img_path.parent.name
            output_name = f"{actor_name}_{img_path.stem}_stylized.png"
            Image.fromarray(stylized_np).save(OUTPUT_STYLIZED / output_name)
        except Exception as e:
            print(f"\n⚠️  Failed to process {img_path.name}: {e}")
    
    print(f"\n✅ Batch processing complete!")
    print(f"📁 All stylized images saved to: {OUTPUT_STYLIZED}")

print("="*60)
```
